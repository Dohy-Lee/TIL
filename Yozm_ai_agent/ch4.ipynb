```python
# DuckDuckgo + Agent
from agents import Agent, Runner, function_tool
from ddgs import DDGS

@function_tool()
def news_search(query:str) -> str:
    """덕덕고를 사용한 뉴스 검색 핸들러 함수"""
    try:
        results = DDGS().text(query, max_results=5)
        if results:
            return f"    '{query}' 검색 결과 :\n{results}"
        else:
            return "검색 결과가 없습니다."

    except Exception as e:
        return f"검색 중 오류가 발생하였습니다: {str(e)}"


news_agent = Agent(
    name="NewsSearchAgent",
    model='gpt-5-mini',
    instructions=(
        "당신은 한국어 뉴스 리포터입니다."
        "WebSearchTool을 사용하여 최신 뉴스를 검색하고,"
        "3개의 기사 URL을 함께 알려주세요."
    ),
    tools=[news_search],
)

if __name__=='__main__':
    print("뉴스 검색 에이전트를 시작합니다.")

    result = await Runner.run(starting_agent=news_agent, input="최신 인공지능 기술 뉴스 검색을 해주세요.")
    print(result.final_output)
```
